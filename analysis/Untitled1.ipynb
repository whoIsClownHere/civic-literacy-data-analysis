```python
import pandas as pd

compare = pd.read_csv("../compare.csv")
compare.head()
```

```python
coders = ["billie", "claude", "chatgpt", "ryan"]


compare["category_difference"] = (
    compare[coders]
    .nunique(axis=1, dropna=True)
    .gt(1)
)


response_comparison = (
    compare.groupby("response", as_index=False)
    .agg(
        difference=("category_difference", "any"),
        number_of_differences=("category_difference", "sum")
    )
)

response_comparison["difference"] = response_comparison["difference"].map({
    True: "difference",
    False: "no difference"
})

response_comparison
```

```python
differing_categories = (
    compare.loc[compare["category_difference"]]
    .groupby("response")["category"]
    .apply(list)
    .rename("differing_categories")
)

response_comparison = response_comparison.merge(
    differing_categories,
    on="response",
    how="left"
)

response_comparison["differing_categories"] = (
    response_comparison["differing_categories"]
    .apply(lambda x: x if isinstance(x, list) else [])
)

response_comparison.head()
```
