```python
import numpy as np
import pandas as pd
```

```python
data_1 = pd.read_excel('CompleteDataWithDemographics_Cleaned.xlsx')
data_1.head()
```

```python
data = pd.read_csv('.csv')
data.head()
```

```python
merged = data_1.join(data, how="inner")
merged
```

```python
import numpy as np, pandas as pd
from scipy.stats import fisher_exact, mannwhitneyu

pro = [c for c in merged.columns if c.startswith('Proponent')]
opp = [c for c in merged.columns if c.startswith('Opponent')]
merged['pro_args']   = merged[pro].sum(1)
merged['opp_args']   = merged[opp].sum(1)
merged['total_args'] = merged['pro_args'] + merged['opp_args']

s1 = merged[merged['Path'].isin(['A(1->4)','D(4->1)'])]
s2 = merged[merged['Path'].isin(['B(2->3)','C(3->2)'])]

def cliffs(a, b):
    a, b = np.asarray(a), np.asarray(b)
    return (sum((x>b).sum() for x in a) - sum((x<b).sum() for x in a)) / (len(a)*len(b))

def fisher_col(c):
    tab = [[s1[c].sum(), (s1[c]==0).sum()], [s2[c].sum(), (s2[c]==0).sum()]]
    OR, p = fisher_exact(tab)
    print(f"{c}: s1={s1[c].mean():.2f} s2={s2[c].mean():.2f} OR={OR:.2f} p={p:.3f}")

def mwu_col(c):
    U, p = mannwhitneyu(s1[c], s2[c], alternative='two-sided')
    print(f"{c}: med s1={s1[c].median()} s2={s2[c].median()} δ={cliffs(s1[c],s2[c]):.2f} p={p:.3f}")
```

```python
rows = []
for c in pro + opp:
    rows.append([c.split(' - ')[1],
                 f"{s1[c].mean()*100:.0f}%",
                 f"{s2[c].mean()*100:.0f}%",
                 f"{merged[c].mean()*100:.0f}%"])

pct_tbl = pd.DataFrame(rows, columns=['category','sample_1','sample_2','all'])
pct_tbl
```

```python
def dumbbell(ax, cols, title):
    p1 = [s1[c].mean() for c in cols]
    p2 = [s2[c].mean() for c in cols]
    labels = [c.split(' - ')[1] for c in cols]
    y = range(len(cols))
    for yi, a, b in zip(y, p1, p2):
        ax.plot([a, b], [yi, yi], color='lightgray', zorder=1)
    ax.scatter(p1, y, color='#1f77b4', label='sample_1 (1↔4)', zorder=2)
    ax.scatter(p2, y, color='#d62728', label='sample_2 (2↔3)', zorder=2)
    ax.set_yticks(list(y)); ax.set_yticklabels(labels)
    ax.set_xlim(0, 1); ax.set_xlabel('Proportion of participants')
    ax.set_title(title); ax.legend()

fig, (axL, axR) = plt.subplots(1, 2, figsize=(13, 4))
dumbbell(axL, pro, 'Proponent')
dumbbell(axR, opp, 'Opponent')
plt.tight_layout(); plt.show()
```

```python
rows = []
for c in pro + opp:
    tab = [[s1[c].sum(), (s1[c]==0).sum()], [s2[c].sum(), (s2[c]==0).sum()]]
    OR, p = mannwhitneyu(tab)
    rows.append([c, round(s1[c].mean(),2), round(s2[c].mean(),2), round(OR,2), round(p,3)])

fisher_tbl = pd.DataFrame(rows, columns=['metric','s1','s2','OR','p']).sort_values('p')
fisher_tbl
```
